```python
# Load env variables and create client
from dotenv import load_dotenv
load_dotenv()

# Create an API client
from anthropic import Anthropic

client = Anthropic()

model = "claude-sonnet-4-6" 
```

```python
def add_user_message(messages, text):
    user_message = {"role":"user", "content":text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role":"assistant", "content":text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
    model=model,
    max_tokens=100,
    messages= messages
    )  
    return message.content[0].text # type: ignore
```

```python
messages = []

add_user_message(messages, "How do I solve 5x + 3=2 for x?")
answer = chat(messages)
answer
```

```python
# let's update the chat function now and see
def chat(messages):
    system = """
    You are a patient math tutor.
    Do not directly answer a studet's questions.
    Guide them to a solution step by step.
    """
    message = client.messages.create(
    model=model,
    max_tokens=100,
    messages= messages,
    system=system
    )  
    return message.content[0].text # type: ignore
```

```python
# let's try it again
messages = []

add_user_message(messages, "How do I solve 5x + 3=2 for x?")
answer = chat(messages)
print(answer)
```

```python
# We are not allow to have a system = to None we have to refactor the chat function to make it generic for other tasks
def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages":  messages
    }
    if system:
        params["system"] = system
    message = client.messages.create(**params)  
    return message.content[0].text # type: ignore
```

```python
messages = []

system = """
    You are a patient math tutor.
    Do not directly answer a studet's questions.
    Guide them to a solution step by step.
    """
add_user_message(messages, "How do I solve 5x + 3=2 for x?")
#answer = chat(messages) # given the update above we can run chat with and without system param
answer = chat(messages, system)
print(answer)
```
